Unlike a laptop or desktop, quantum computers have no keyboard or screen and require remote access.

A Quantum Processing Unit (QPU) is made from quantum mechanical objects such as superconducting qubits, trapped ions, neutral atoms, and photons.

The QPU is sensitive to interactions with the wider world which make the qubits lose their quantum properties (*decohere*). Therefore, using a quantum computer is completely different to how we use everyday classical computers.

# From Bits to Qubits

## Quantum States

Classical computers operate using *bits*, which take values of either $0$ or $1$.

Quantum computers operate using *qubits* (quantum bits/a quantum system made of two energy levels)
<p align="center">
    <img src="https://upload.wikimedia.org/wikipedia/commons/c/cc/Zeemn_effect_demonstr.png?utm_source=commons.wikimedia.org&utm_campaign=index&utm_content=original" width="400"/>
</p>

The qubit state is described by a quantum statevector, $|\psi\rangle$.

A general quantum state of a qubit can be written as
\begin{equation}
  |\psi\rangle = a\begin{pmatrix}{1 \\ 0}\end{pmatrix} + b\begin{pmatrix}{0 \\ 1}\end{pmatrix} = a|0\rangle + b|1\rangle, a,b\in\mathbb{C}
\end{equation}

For $|\psi\rangle$ to be a quantum state rather than a generic complex vector, it must satisfy
\begin{equation}
  |a|^2+|b|^2=1
\end{equation}

When a quantum state is measured, it can only return $0$ or $1$, not the full state.\
$\implies$ You can only extract classical information from quantum information.

\begin{align}
  P(0) = |a|^2\\
  P(1) = |b|^2
\end{align}

The key condition for a vector to be a quantum state is equivalent to all probabilities must sum to $1$.

Notice that a quantum state can be a combination of the two classical states, existing in both at the same time.

This property is called *quantum superposition*.

Let's visualise what a qubit state can look like.

```python
import numpy as np # Importing numpy
from qiskit.quantum_info import Statevector # Importing the Qiskit Statevector class
ket0 = [[1],[0]] # Defining the elements of the vector
psi = Statevector(ket0) # Defining the vector as a Statevector class object
print(psi.is_valid()) # Checking that the statevector satisfies the properties of a quantum state

from qiskit.visualization import plot_bloch_multivector # Importing the function to plot on the Bloch Sphere
plot_bloch_multivector(psi) # Plotting on the Bloch Sphere
```

## Quantum Operators

With the knowledge of what a quantum state is, how can one quantum state be changed into another?

The answer is with a quantum operator $O$, such that
\begin{equation}
  O|\psi\rangle = |\phi\rangle
\end{equation}
where $|\phi\rangle$ is another quantum state.

If $O$ changes the quantum state from one 2-dimensional vector to another, then it must be a $2\times 2$ matrix,

\begin{equation}
  O = \begin{pmatrix} o_{11} & o_{12}\\ o_{21} & o_{22} \end{pmatrix}
\end{equation}

There are two main types of quantum operator,


*   An operator $O^\dagger=O$ is *hermitian* and corresponds to something we can observe, e.g. position, momentum, electric charge.
*   An operator $O^\dagger=O^{-1}$ is *unitary* and corresponds to an operation which is reversible.

Note that the *hermitian conjugate* of an operator $O$ is

\begin{equation}
  O^\dagger = \begin{pmatrix} o_{11}^* & o_{21}^* \\ o_{12}^* & o_{22}^* \end{pmatrix}
\end{equation}

Let's look at some of the most common quantum operations.



```python
from qiskit.quantum_info.operators import Operator # Importing the Qiskit Operator class
from qiskit.circuit.library import XGate # Importing the X-gate from Qiskit's circuit library

# Defining an initial |0> state
ket0 = [[1],[0]]
psi = Statevector(ket0)

X = XGate()
X_op = Operator(X) # Converting the X-gate to an object from the Operator class
print(X_op.is_unitary()) # Checking if the X-operation is unitary

psi_new = np.dot(X_op,psi) # Acting on the initial state with X
print(psi_new)
plot_bloch_multivector(psi_new) # Visualising the result
```

We see that the $X$ operator flips the $|0\rangle$ state to a $|1\rangle$ state.

## Quantum Circuits

Everything to this point has just been linear algebra. Where does the quantum computer appear?

Everything done thus far can also be carried out using *quantum circuits*.

Quantum circuits more closely resemble circuits from computer science rather than physics.\
The collection of qubits in a quantum circuit is called a *quantum register*, and the quantum operations are called *quantum gates*.

All qubits in the qubit register begin in the $|0\rangle$ state.

```python
from qiskit import QuantumCircuit # Importing the QuantumCircuit class

qc = QuantumCircuit(1) # Defining a quantum circuit with a single qubit
qc.draw('mpl') # Drawing the circuit
```

Let's do each of the previous quantum operation examples as quantum circuits.

```python
# Ex 1: X-gate on |0>
qc1 = QuantumCircuit(1)
qc1.x(0) # Applying an X-gate to qubit 0 (Note: Indices start counting from 0)
display(qc1.draw('mpl'))

# Ex 2: H-gate on |0>
qc2 = QuantumCircuit(1)
qc2.h(0) # Applying a H-gate to qubit 0
display(qc2.draw('mpl'))

# Ex 3: H-gate then a Z-gate on |0>
qc3 = QuantumCircuit(1)
qc3.h(0) # Applying a H-gate to qubit 0
qc3.z(0) # Applying a Z-gate to qubit 0
qc3.draw('mpl')
```

Now it is clear how operations are applied, but the resulting quantum state has to be found from the circuit.

```python
# Ex 1:
state1 = Statevector.from_circuit(qc1)

# Print or draw the complex amplitudes
print(state1.data)
display(state1.draw("latex"))

# Ex 2:
state2 = Statevector.from_circuit(qc2)

# Print or draw the complex amplitudes
print(state2.data)
display(state2.draw("latex"))

# Ex 3:
state3 = Statevector.from_circuit(qc3)

# Print or draw the complex amplitudes
print(state3.data)
display(state3.draw("latex"))
```

However, in any case where a quantum computer would have benefit beyond classical computation, we cannot simply apply the gates ourselves and know what state should appear at the end.

All we can do on real quantum computers is measure the qubits.

```python
from qiskit_aer import AerSimulator # Importing the Qiskit AerSimulator to run quantum circuits locally
from qiskit_ibm_runtime import SamplerV2 as Sampler # Importing the Sampler class to measure qubit counts
from qiskit.visualization import plot_histogram # Importing the function to plot histograms

backend = AerSimulator() # Initialising a backend

# Take the circuit from Ex 2
qc_meas = qc2.copy() # Copying the original circuit
qc_meas.measure_all() # Adding a measurement gate and classical bit to store the result
display(qc_meas.draw('mpl')) # Drawing the circuit

sampler = Sampler(mode=backend) # Initialising the sampler on the selected backend
sampler_result = sampler.run([qc_meas], shots=1000).result() # Running the circuit and measurin 1000 times
counts = sampler_result[0].data.meas.get_counts() # Extracting the measurement counts for each result

display(plot_histogram(counts)) # Displaying the results
```

## Multiple Qubits

While a single qubit can exhibit genuinely quantum behaviour, a classical computer can calculate anything a single qubit quantum computer can. The real benefit comes from quantum registers of many qubits.

$|\psi\rangle$ can no longer be written as a vector with $2$ complex entries. For a collection of $N$ qubits, $|\psi\rangle$ requires $2^N$ complex entries.\
$\implies$ When $N$ gets very large classical computers no longer even have the memory to store $|\psi\rangle$.

$|\psi\rangle$ can now be written as
\begin{equation}
  |\psi\rangle = \sum_{i=0}^{2^N-1}c_i|i\rangle
\end{equation}

e.g. $N=1$, $|i\rangle\in\{|0\rangle,|1\rangle\}$.\
e.g. $N=2$, $|i\rangle\in\{|00\rangle,|01\rangle,|10\rangle,|11\rangle\}$.

The state for an $N$ qubit system can be built from single qubit states using the *tensor product*.\
For two states, $|a\rangle=\sum_ia_i|i\rangle$ and $|b\rangle=\sum_jb_j|j\rangle$, their tensor product is

\begin{equation}
  |a\rangle\otimes|b\rangle=\sum_{i,j}a_ib_j|i\rangle\otimes|j\rangle
\end{equation}

In more familiar column vector notation this looks like

\begin{equation}
  \begin{pmatrix}a_0 \\ \vdots \\ a_{2^{N_1}-1}\end{pmatrix} \otimes \begin{pmatrix}b_0 \\ \vdots \\ b_{2^{N_2}-1}\end{pmatrix} = \begin{pmatrix} a_0b_0 \\ \vdots \\ a_0b_{2^{N_2}-1} \\ \vdots \\ a_{2^{N_1}-1}b_0 \\ \vdots \\ a_{2^{N_1}-1}b_{2^{N_2}-1}\end{pmatrix}
\end{equation}

Notice the use of $N_1$ and $N_2$. The tensor product can be taken between any pair of vectors regardless of their lengths.

The tensor product of operators is then

\begin{aligned}
  A \otimes B &= \begin{pmatrix}a_{11} & \dots & a_{1,2^{N_1}-1}\\ \vdots & \dots & \vdots\\ a_{2^{N_1}-1,1} & \dots & a_{2^{N_1}-1,2^{N_1}-1}\end{pmatrix} \otimes \begin{pmatrix} b_{11} & \dots & b_{1,2^{N_2}-1}\\ \vdots & \dots & \vdots\\ b_{2^{N_2}-1,1} & \dots & b_{2^{N_2}-1,2^{N_2}-1}\end{pmatrix}\\
  &= \begin{pmatrix} a_{11}B & \dots & a_{1,2^{N_1}-1}B \\ \vdots & \dots & \vdots \\ a_{2^{N_1}-1,1}B & \dots & a_{2^{N_1}-1,2^{N_1}-1}B \end{pmatrix}
\end{aligned}

So the tensor product of any objects of size $P\times Q$ and $K\times L$ has resulting size $PK\times QL$.

The tensor product now lets us define *local* vs *global* quantum gates.\
A local quantum gate is an operator that can be decomposed as a tensor product over the qubits
\begin{equation}
  O_{\text{global}} = \mathbb{I}\otimes\dots\otimes O \otimes\dots\mathbb{I}
\end{equation}
whereas a global quantum gates acts on all qubits collectively and cannot be decomposed in this way.

The tensor product also enables us to define another property unique to quantum systems, *quantum entanglement*.\
Two qubits, $A$ and $B$, are said to be entangled if their global state $|\psi\rangle_{AB}$ <ins>cannot</ins> be written as $|\psi\rangle_{A}\otimes|\psi\rangle_{B}$.

This means that when two quantum systems are entangled, neither has an individually well-defined quantum state, only their global quantum state can be said to be well-defined.

The most famous entangled states are the Bell states
\begin{aligned}
  |\Phi^+\rangle &= \frac{|00\rangle+|11\rangle}{\sqrt{2}}\\
  |\Phi^-\rangle &= \frac{|00\rangle-|11\rangle}{\sqrt{2}}\\
  |\Psi^+\rangle &= \frac{|01\rangle+|10\rangle}{\sqrt{2}}\\
  |\Psi^-\rangle &= \frac{|01\rangle-|10\rangle}{\sqrt{2}}
\end{aligned}

Let's look at the quantum circuit to make $|\Phi^+\rangle$.

```python
qc_bell = QuantumCircuit(2)
qc_bell.h(0)
qc_bell.cx(0,1) # Applying the 2-qubit CNOT gate
display(qc_bell.draw('mpl'))

state_bell = Statevector.from_circuit(qc_bell)
display(state_bell.draw("latex"))
```

# Real Quantum Computing

Can everything be done ideally, with no knowledge of the quantum computer itself?\
The answer is <ins>no</ins>, simulations of current quantum computers require knowledge of the connectivity of the QPU, the operations the quantum computer can perform, and common sources of error that exist at the hardware level.

## Basis Gates and Transpilation

To get the most accurate result from a simulation on any quantum computer a quantum circuit must be efficiently mapped onto the best choice qubits on the QPU, and the quantum gates expressed using the *basis gates* of the quantum computer.

This process is called *transpilation*.

Let's take a look at a QPU on one of IBM's quantum computers.

```python
from qiskit.visualization import plot_gate_map # Importing the function to plot the QPU layout
from qiskit_ibm_runtime.fake_provider import FakeMarrakesh # Importing the info on a real IBM QPU

backend_qpu = FakeMarrakesh() # Initialising the backend
plot_gate_map(backend_qpu) # Plotting the QPU
```

Notice that not all qubits connect to each other. When running a quantum circuit the goal would be for any qubits that interact through a $2$-qubit gate to be next to each other.

Let's see what operations this QPU can perform.

```python
print(backend_qpu.operation_names)
```

Notice that the previously used Hadamard and $Z$ gates are not in this set. This means that they cannot be applied directly on this QPU, and instead must be built from the existing gates.

So how can all of this be handled in Qiskit?

Step 1: Create the desired quantum circuit.

Let's create a $3$-qubit entangled GHZ-state.
\begin{equation}
  |\psi⟩ = \frac{|000⟩+|111⟩}{\sqrt{2}}
\end{equation}

```python
qc_ghz = QuantumCircuit(3)

qc_ghz.h(0)
qc_ghz.cx(0,1)
qc_ghz.cx(1,2)
qc_ghz.measure_all()
display(qc_ghz.draw('mpl'))
```

Step 2: Map the desired quantum circuit to the chosen QPU through built-in Qiskit functions.

```python
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager

pm = generate_preset_pass_manager(optimization_level=1,backend=backend_qpu) # Creating a pass-manager with the QPU info
isa_qc_ghz = pm.run(qc_ghz) # Running the quantum circuit through the pass-manager to create an Instruction-Set-Architecture circuit

isa_qc_ghz.draw('mpl') # Drawing the transpiled circuit
```

Notice that the total number of gates has now significantly increased.

Step 3: Run the ISA circuit and expect to get the same result as the desired circuit.

```python
sampler = Sampler(mode=backend_qpu) # Initialising the sampler on the selected backend
sampler_result = sampler.run([isa_qc_ghz], shots=1000).result() # Running the circuit and measurin 1000 times
counts = sampler_result[0].data.meas.get_counts() # Extracting the measurement counts for each result

display(plot_histogram(counts)) # Displaying the results
```

The measured bitstrings show $000$ and $111$ with near equal probability as expected. However, other bitstrings that don't appear in $|\psi⟩$ are also measured.

These unexpected measurements do not come from any errors in the transpilation process, but are instead errors from the quantum computer itself.

## Errors

When using our everyday classical computers, they are sufficiently developed to the point where any error in a calculation arises due to user/method errors. We never consider any errors at the level of bits.

The typical range for a bit error on a classical computer is $10^{-15}-10^{-17}$ (roughly one error per $10^{15}-10^{17}$ transistor switches).

For modern day qubits the error range would be $10^{-2}-10^{-3}$ (an error every $10^{2}-10^{3}$ quantum gates).



```python
from qiskit.visualization import plot_error_map # Importing the function to plot the error map
plot_error_map(backend_qpu, figsize=(25,20)) # Plotting the error map
```

Unlike classical computers, quantum computers have a larger range of errors due to the larger state space that qubits possess. Beyond bit-flip errors, qubits can also experience

*   Phase-Flip Errors: A change of relative phase between quantum state ($|0⟩+|1⟩→|0⟩-|1⟩$).
*   Bit and Phase-Flip Errors: A simultaneous change in phase and qubit state ($|0⟩→i|1⟩,|1⟩→-i|0⟩$).
*   Rotation Errors: An over/under-rotation of the qubit state on the Bloch Sphere, which leaves a valid but incorrect quantum state. Unlike in classical computation where the change in state is discrete ($0↔1$).
*   Relaxation Errors: A spontaneous loss of energy from the qubit and a resulting drop from the $|1⟩$ state to $|0⟩$.
*   Dephasing Errors: A loss in phase coherence between the qubit states.
